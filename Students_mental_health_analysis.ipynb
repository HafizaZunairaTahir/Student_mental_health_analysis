### 🔎 Observation Set 3:
---
After auditing the unique values in the categorical columns, several Data Quality Issues were identified:

**Casing Inconsistency:** The year column contains both 'year 1' and 'Year 1'. Python treats these as two different categories, which will split our data inaccurately in charts.

**Redundant Categories:** In the course column, we see 'Engineering', 'Engine', and 'engin'. These are clearly the same department but entered differently.

**Whitespace Noise:** Some entries have "invisible" spaces at the end (e.g., 'Accounting '), which prevents the data from grouping correctly.

**Language Variation:** 'Pendidikan Islam' and 'Islamic Education' represent the same subject. To get an accurate count of student stress by course, these must be merged.

>**Decision:** We must perform Text Normalization to ensure our Phase 5.0 (Visualization) results are statistically sound.

```python
# FINAL DATA NORMALIZATION

# 1. Fixing the 'year' column
# We strip spaces and make everything lowercase so 'Year 1' and 'year 1' match
df['year'] = df['year'].str.strip().str.lower()
```

```python
# 2. Fixing the 'course' column
# We strip spaces and make everything uppercase for a professional look
df['course'] = df['course'].str.strip().str.upper()
```

```python
# 3. Handling course abbreviations (Example: ENGINE and ENGIN -> ENGINEERING
course_mapping = {
    'ENGINE': 'ENGINEERING',
    'ENGIN': 'ENGINEERING',
    'BIT': 'IT',
    'PENDIDIKAN ISLAM': 'ISLAMIC EDUCATION',
    'PENDIDIKAN ISLAM ': 'ISLAMIC EDUCATION',
    'PSYCHOLOGY': 'PSYCHOLOGY',
    'LAWS': 'LAW'
}
df['course'] = df['course'].replace(course_mapping)
```

```python
# 4. Verification
print(f"Cleaned Years: {df['year'].unique()}")
print(f"Total Unique Courses after merging: {df['course'].nunique()}")
```

### Observation Set 4:
---
Before proceeding to Bivariate Analysis and Correlation mapping, we must transform our categorical text into numerical formats.

**Why this is necessary:**
* **Mathematical Operations:** To calculate correlations between 'Age' and 'Mental Health', both variables must be numeric.
* **Ordinal Sorting:** Converting CGPA ranges (e.g., '3.0 - 3.49') into a numeric scale (0-4) ensures that our visualizations follow a logical progression from lowest to highest achievement.
* **Boolean Mapping:** Converting 'Yes/No' responses into '1/0' allows us to calculate the "density" of mental health issues across different student groups.

> **Note:** The original categorical columns are preserved for labeling purposes, while the new `_numeric` and `_bool` columns will be used for statistical computation.

## 🔵 5.2 BIVARIATE ANALYSIS
In this phase, we explore the relationships between two different variables. By comparing demographics against mental health indicators, we can identify specific "High-Risk" groups.

### 🎯 Research Objectives:
* **Correlation Mapping:** Using a heatmap to see how age, year of study, and CGPA relate to mental health.
* **Academic Stress:** Investigating if higher CGPA (Toppers) correlates with higher anxiety.
* **Trend Analysis:** Determining if mental health issues intensify as students progress from Year 1 to Year 4.

```python
# CONVERTING CATEGORICAL TO NUMERIC (FOR BIVARIATE ANALYSIS)

# 1. Extracting numbers from 'year' (e.g., 'year 1' -> 1)
df['year_numeric'] = df['year'].str.extract('(\d+)').astype(int)
```

```python
# 2. Encoding CGPA ranges into a numeric scale
# This ensures that 0-1.99 comes BEFORE 3.50-4.00 in our analysis
cgpa_map = {
    '0 - 1.99': 0,
    '2.00 - 2.49': 1,
    '2.50 - 2.99': 2,
    '3.00 - 3.49': 3,
    '3.50 - 4.00': 4
}
df['cgpa_numeric'] = df['cgpa'].map(cgpa_map)
```

```python
# 3. Converting Yes/No to 1/0 (Essential for Heatmaps/Correlations)
for col in ['depression', 'anxiety', 'panic_attack', 'treatment']:
    df[f'{col}_bool'] = df[col].map({'Yes': 1, 'No': 0})

print(df[['year', 'year_numeric', 'cgpa', 'cgpa_numeric']].head())
```

```python
# Fig-5: Correlation Heatmap
plt.figure(figsize=(12, 8))

# We only use the numeric and boolean columns that we created 
corr_cols = ['age', 'year_numeric', 'cgpa_numeric', 'depression_bool', 'anxiety_bool', 'panic_attack_bool']
correlation_matrix = df[corr_cols].corr()

# Plotting the Heatmap
sns.heatmap(correlation_matrix, 
            annot=True, 
            cmap='Blues', 
            fmt=".2f", 
            linewidths=0.5,
            cbar_kws={'label': 'Correlation Coefficient'})

plt.title('Fig-5: Correlation Heatmap of Student Features', fontsize=16, fontweight='bold', color='#000080', pad=20)
plt.show()
```

>**Fig-5**: Correlation Heatmap of Student Features

The heatmap provides a mathematical summary of how variables interact within our dataset. Key observations include:

* **Mental Health Co-occurrence (0.27):** There is a notable positive correlation between **Depression** and **Anxiety**, suggesting these conditions often manifest together.
* **Academic Pressure (0.18):** A positive correlation exists between **CGPA** and **Anxiety**, supporting the hypothesis that high academic achievement may be linked to increased stress.
* **The Seniority Paradox (-0.02):** Interestingly, there is almost zero correlation between the **Year of Study** and mental health indicators, suggesting that stress is prevalent across all academic stages, not just the final year.

```python
# Preparing the timeline data
# We calculate the average anxiety 'density' for every Age/Year combination
timeline_df = df.groupby(['age', 'year']).agg(
    student_count=('anxiety', 'count'),
    anxiety_rate=('anxiety_bool', 'mean') # This gives us a 0 to 1 score of how common anxiety is
).reset_index()

# Fig-6: The Student Life Mental Health Timeline
fig_timeline = px.scatter(timeline_df, 
                         x="age", 
                         y="year",
                         size="student_count", 
                         color="anxiety_rate",
                         hover_data={'anxiety_rate': ':.2f', 'student_count': True},
                         title="<b>Fig-6: Mental Health Timeline: Age vs. Academic Progression</b>",
                         labels={'anxiety_rate': 'Anxiety Density', 'age': 'Student Age (Timeline)'},
                         color_continuous_scale='RdBu_r', # Red for high anxiety, Blue for low
                         template="plotly_white")

# Making it look like a professional timeline
fig_timeline.update_layout(
    title_font_color="#000080",
    xaxis=dict(title="Age Timeline", showgrid=False, dtick=1),
    yaxis=dict(title="Academic Year", showgrid=True),
    height=500
)

fig_timeline.show()
```

>**Fig-6**: Mental Health Timeline: Age vs. Academic Progression.

This bubble chart acts as a "Life Map" of the students in our dataset. By plotting **Age** against **Year of Study**, we can observe the following:

1. **Concentration of Stress:** Larger, darker bubbles represent "Hotspots" where a high volume of students are reporting anxiety. 
2. **The Freshman Peak:** In this dataset, we see significant density in the **18-19 age bracket (Year 1)**, suggesting that the transition into university life is a primary trigger for mental health struggles.
3. **The Maturity Gap:** As we move along the X-axis (Age), the bubbles become more sparse. This indicates that while older students are fewer in number, their "Anxiety Density" (color) remains a factor worth investigating.
4. **Multivariate Insight:** By using size for student count and color for anxiety, we have successfully visualized three dimensions of data in a single view.

## 🔵 5.3 MULTIVARIATE ANALYSIS
To conclude our visualization phase, we will perform a bulk comparison of mental health indicators across genders. This allows us to see which condition is most prevalent in specific demographic groups.

```python
# 1. Selecting the columns and melting the data for a bulk plot
mh_cols = ['depression', 'anxiety', 'panic_attack', 'gender']
melted_df = df[mh_cols].melt(id_vars='gender', var_name='Condition', value_name='Status')

# 2. Filtering only 'Yes' responses to see the prevalence
melted_df_yes = melted_df[melted_df['Status'] == 'Yes']

# 3. Plotting the Bulk Comparison
plt.figure(figsize=(14, 8))
sns.countplot(data=melted_df_yes, x='Condition', hue='gender', palette=['#000080', '#4682B4'])

# 4. Styling for a "Board Report" look
plt.title('Fig-7: Bulk Comparison of Mental Health Issues by Gender', fontsize=16, fontweight='bold', color='#000080', pad=20)
plt.xlabel('Mental Health Condition', fontsize=12)
plt.ylabel('Number of Reported Cases', fontsize=12)
plt.legend(title='Gender', frameon=True)
plt.grid(axis='y', linestyle='--', alpha=0.4)

# Adding the counts on top
for p in plt.gca().patches:
    plt.gca().annotate(f'{int(p.get_height())}', (p.get_x() + p.get_width() / 2., p.get_height()), 
                       ha='center', va='center', xytext=(0, 8), textcoords='offset points', fontweight='bold')

plt.show()
```

>**Fig-7**: Bulk Comparison of Mental Health Issues by Gender.

This bulk visualization reveals the distribution of psychological struggles across the gender divide:

1. **Predominant Condition:** We can clearly identify which condition (Anxiety, Depression, or Panic Attacks) has the highest "peak" for both groups.
2. **Gender Disparity:** The chart highlights how much more frequently **Females** report these issues compared to **Males** in this specific dataset.
3. **Symmetry of Struggles:** (Look at your chart)—Do the bars follow the same pattern for both genders, or does one gender struggle more with Anxiety while the other struggles with Depression?

### Q1: Identifying students suffering from Depression, Anxiety, AND Panic Attacks

```python
triple_threat_df = df[(df['depression'] == 'Yes') & 
                      (df['anxiety'] == 'Yes') & 
                      (df['panic_attack'] == 'Yes')]

total_count = len(df)
triple_count = len(triple_threat_df)
percentage = (triple_count / total_count) * 100

print(f"Total Student Sample: {total_count}")
print(f"Students in 'Triple Threat' Zone: {triple_count}")
print(f"Percentage of High-Risk Students: {percentage:.2f}%")

# Visualize the finding
plt.figure(figsize=(8, 5))
sns.barplot(x=['Total Students', 'High-Risk (All 3 Conditions)'], 
            y=[total_count, triple_count], 
            palette=['#cbd5e1', '#ff4d4d'])

plt.title('Q1: Magnitude of the "Triple Threat" Group', fontweight='bold', color='#000080')
plt.ylabel('Number of Students')
plt.show()
```

### 💡 Investigation 1 Result: The Triple Threat
Our query reveals that **9.9%** of the student population is battling Depression, Anxiety, and Panic Attacks simultaneously. 

* **Critical Insight:** This group represents the highest-priority individuals for university mental health interventions. 
* **Observation:** While the individual percentages for each condition might seem manageable, seeing the overlap highlights the severity of the psychological burden on a specific subset of students.

### Q5: Do older students (Age 22-24) show different mental health patterns compared to younger students (Age 18-19)?


```python
# Create Age Groups
def age_group(age):
    if age <= 19: return 'Younger (18-19)'
    elif age >= 22: return 'Older (22-24)'
    else: return 'Mid-range'

df['age_category'] = df['age'].apply(age_group)
age_stability = df.groupby('age_category')['anxiety_bool'].mean() * 100

plt.figure(figsize=(8, 5))
age_stability.loc[['Younger (18-19)', 'Older (22-24)']].plot(kind='bar', color='teal')
plt.title('Q5: Anxiety Rates: Younger vs Older Students', fontweight='bold')
plt.ylabel('Anxiety Rate (%)')
plt.xticks(rotation=0)
plt.show()
```

### 💡 Investigation 5 Result: The Age vs. Stability Factor
Comparing age extremes reveals that **Younger Students (18-19)** experiences higher anxiety rates than their counterparts.

* **Critical Insight:** Age is a defining factor in coping mechanisms. Younger students may struggle with the transition to independence, while older students face career-entry anxiety.
* **Observation:** Mental health support should not be "one size fits all" but should instead be age-specific.
